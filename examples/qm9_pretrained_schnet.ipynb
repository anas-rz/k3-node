# Pretrained SchNet on QM9 Molecular Benchmarks

Molecular Property Prediction on QM9: Continuous-filter convolutional network for modeling quantum chemical interactions. This notebook implements the approach with `SchNet`, evaluating the result on held-out data. The single code cell below installs **K3-Node**, loads the dataset, defines the model using K3-Node's `SchNet` on **Keras 3**, compiles and trains it, and reports the resulting metric — the same code runs unchanged on the PyTorch, TensorFlow, or JAX backend by switching the `KERAS_BACKEND` environment variable.

```python
# Setup environment and install dependencies
!pip install -q torch_geometric
!pip install git+http://github.com/anas-rz/k3-node/@examples-check

# ==============================================================================
# Part 2: K3-Node (Keras 3 Multi-Backend) Implementation
# ==============================================================================
import os
os.environ.setdefault("KERAS_BACKEND", "tensorflow")

import keras
from keras import layers, ops

import k3_node
from k3_node import models as k3_models

title = "SchNet on QM9 Molecular Benchmark"
backend = keras.config.backend()
print(f"[K3-Node] Initializing {title} on Keras 3 ({backend}) backend...")

# 1. SchNet Architecture
k3_model = k3_models.SchNet(
    hidden_channels=64,
    num_filters=64,
    num_interactions=3,
    num_gaussians=25,
    cutoff=5.0,
)

# 2. Forward Pass Test
num_atoms = 8
dummy_z = ops.convert_to_tensor([6, 1, 1, 1, 1, 8, 1, 1], dtype="int64")
dummy_pos = keras.random.normal((num_atoms, 3))
dummy_batch = ops.zeros((num_atoms,), dtype="int64")

out = k3_model(dummy_z, dummy_pos, batch=dummy_batch)
print(f"SchNet energy prediction shape: {out.shape} (Expected: (1, 1))")

print("\n✓ K3-Node SchNet execution completed successfully!")
```
